```python
# Import all depedencies

from langchain_openai import ChatOpenAI
from langgraph_supervisor.supervisor import create_supervisor
from langgraph.prebuilt import create_react_agent
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_community.tools import YouTubeSearchTool
```

```python
# Set up OpenAI API Key
# Please make sure you have your OpenAI API key stored as a secret named 'OPENAI_API_KEY'
import os
from google.colab import userdata

os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_APIKEY')
```

```python
# ==========================================
# Step 1: Initialize LLM
# ==========================================
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
# ==========================================
# Step 2: Setup DuckduckGo
# ==========================================

web_search_tool = DuckDuckGoSearchRun()

# Example usage
result = web_search_tool.invoke("What is MicroService Architecture?")
print(result)
```

```python
# ==========================================
# Step 3: Setup YouTubeSearch
# ==========================================


youtube_search_tool = YouTubeSearchTool()

# Example usage
result = youtube_search_tool.invoke("What is Module Federation micro-frontends?")
print(result)
```

```python
# ==========================================
# Step 4: Setup Text-Agent
# ==========================================

web_search_agent = create_react_agent(
    model=llm,
    tools=[web_search_tool],
    name="web_search_expert",
    prompt="You are a web research expert. Search the web for non-video textual information."
)
```

```python
# ==========================================
# Step 5: Setup Video-Agent
# ==========================================

youtube_search_agent = create_react_agent(
    model=llm,
    tools=[youtube_search_tool],
    name="youtube_search_expert",
    prompt="You are a YouTube search expert. Retrieve relevant video titles and links from YouTube."
)
```

```python
supervior_agent = create_supervisor(
    agents=[web_search_agent, youtube_search_agent],
    model=llm,
    prompt=(
        "You are a team supervisor coordinating research requests. "
        "Delegate questions about video content, tutorials, or streams to 'youtube_search_expert'. "
        "Delegate factual web lookups, news, or text documentation to 'web_search_expert'."
    )
)
```

```python
# ==========================================
# Step 6: Run Supervisor-Agent
# ==========================================

agentExecutor = supervior_agent.compile();

results = agentExecutor.invoke({
    "messages": [
        {"role": "user", "content": "Find me YouTube videos explaining LangGraph multi-agent systems, and search the web for its documentation."}
    ]
})

print(results)
```

```python
# Print final messages
for message in results["messages"]:
    print (message.pretty_print())
```
