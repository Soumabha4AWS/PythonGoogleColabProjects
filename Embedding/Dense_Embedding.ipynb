Dense Embeddings

```
# Import the TextEmbedding
from fastembed import TextEmbedding
```

```
# Show all Embedding Models avilable

for model in TextEmbedding.list_supported_models():
    print(f"Model: {model['model']}")
    print(f"Dimension: {model.get('dim', 'N/A')}")
    print(f"Description: {model.get('description', 'N/A')}")
    print("-" * 40)
```

```
# Show all Embedding Models avilable in Panda Dataframe

import pandas as pd
# Load supported models into a DataFrame
df_models = pd.DataFrame(TextEmbedding.list_supported_models())

# Display the table
print(df_models[["model", "dim", "description"]])
```

```
# Setting up the Dense Embedding Model

model = TextEmbedding(model_name="BAAI/bge-small-en-v1.5")
```

```
# Document that need to be embeeded

documents = [
    "FastEmbed is lightweight and fast.",
    "It is designed for efficient vector generation on CPUs."
]
```

```
# Generate the Embedding

embeddings = list(model.embed(documents))
```

```
# Show the Embedding

for emb in embeddings:
    print(emb.shape)
    print(emb)
```
