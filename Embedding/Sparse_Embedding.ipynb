```
# Import the SparseTextEmbedding
from fastembed import SparseTextEmbedding
```

```
# Show all Embedding Models avilable

for model in SparseTextEmbedding.list_supported_models():
    print(f"Model: {model['model']}")
    print(f"Dimension: {model.get('dim', 'N/A')}")
    print(f"Description: {model.get('description', 'N/A')}")
    print("-" * 40)
```

```
# Show all Embedding Models avilable in Panda Dataframe

import pandas as pd

# Load supported sparse models into a DataFrame
df_sparse_models = pd.DataFrame(SparseTextEmbedding.list_supported_models())

# Display key columns
print(df_sparse_models[["model", "vocab_size", "size_in_GB", "description"]])
```

```
# Setting up the Dense Embedding Model

model = SparseTextEmbedding(model_name="prithivida/Splade_PP_en_v1")
```

```
# Document that need to be embeeded

documents = [
    "FastEmbed is lightweight and fast.",
    "It is designed for efficient vector generation on CPUs."
]
```

```
# Generate the Embedding

embeddings = list(model.embed(documents))
```

```
# Show the Embedding

for emb in embeddings:
    print("Token Indices:", emb.indices[:5])
    print("Token Weights:", emb.values[:5])
    print(emb)
```
