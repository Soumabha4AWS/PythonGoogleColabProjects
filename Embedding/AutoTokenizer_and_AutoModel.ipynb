# AutoTokenizer from Transformer

AutoTokenizer is one of the most important classes in the Transformers library. It is responsible for converting raw text into tokens (numbers) that an LLM or embedding model can understand. Without a tokenizer, an AI model cannot process human-readable text.

            Input Text
                │
                ▼
         AutoTokenizer
                │
      Token IDs + Attention Mask
                │
                ▼
       Transformer Model
                │
                ▼
        Embeddings / Predictions

```
# Import the AutoTokenizer
from transformers import AutoTokenizer, AutoModel
```

```
# use the bert-base-uncased Tokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

print("====" * 80)
text = "FastEmbed is amazing."
real_tokens_pt = tokenizer(text,return_tensors="pt")
# print(f"real_tokens => {real_tokens}")
print(f"input_ids in real_tokens with pt tensor => {real_tokens_pt.input_ids}")
print(f"token_type_ids in real_tokens with pt tensor => {real_tokens_pt.token_type_ids}")
print(f"attention_mask in real_tokens with pt tensor => {real_tokens_pt.attention_mask}")
print("----" * 80)
real_tokens_np = tokenizer(text,return_tensors="np")
# print(f"real_tokens => {real_tokens}")
print(f"input_ids in real_tokens with np tensor => {real_tokens_np.input_ids}")
print(f"token_type_ids in real_tokens with np tensor => {real_tokens_np.token_type_ids}")
print(f"attention_mask in real_tokens with np tensor => {real_tokens_np.attention_mask}")
print("====" * 80)

actual_tokens = tokenizer.tokenize(
    "FastEmbed is amazing."
)
print(f"actual_tokens => {actual_tokens}")
```

```
# use the sentence-transformers all-MiniLM-L6-v2 Tokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "sentence-transformers/all-MiniLM-L6-v2"
)

text = "FastEmbed is amazing."
real_tokens = tokenizer(text,return_tensors="pt")
# print(f"real_tokens => {real_tokens}")
print(f"input_ids in real_tokens => {real_tokens.input_ids}")
print(f"token_type_ids in real_tokens => {real_tokens.token_type_ids}")
print(f"attention_mask in real_tokens => {real_tokens.attention_mask}")

actual_tokens = tokenizer.tokenize(
    "FastEmbed is amazing."
)
print(actual_tokens)
```

```
# use the sentence-transformers all-MiniLM-L6-v2 Model

model = AutoModel.from_pretrained(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```
# Pass tokens to the model

embedding = model(**real_tokens)
```

```
# Show the Embedding

print(embedding.last_hidden_state.shape)
print(embedding.last_hidden_state)
```
